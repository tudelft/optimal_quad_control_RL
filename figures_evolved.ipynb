```python
from matplotlib.patches import FancyArrowPatch
# DEFINE RACE TRACK
r = 1.50
gate_pos = np.array([
    [ r,  -r, -1.5],
    [ 0,   0, -1.5],
    [-r,   r, -1.5],
    [ 0, 2*r, -1.5],
    [ r,   r, -1.5],
    [ 0,   0, -1.5],
    [-r,  -r, -1.5],
    [ 0,-2*r, -1.5]
])
gate_yaw = np.array([1,2,1,0,-1,-2,-1,0])*np.pi/2
start_pos = gate_pos[0] + np.array([0,-1.,0])

fig1 = plt.figure(figsize=(3, 6))
ax1 = fig1.add_subplot(111)

for gate, yaw in zip(gate_pos, gate_yaw):
    gate_line = np.array([[-0.75, 0.0], [0.75, 0.0]])
    gate_line = (np.array([[np.cos(yaw+np.pi/2), -np.sin(yaw+np.pi/2)],[np.sin(yaw+np.pi/2),  np.cos(yaw+np.pi/2)]]) @ gate_line.T).T
    # ax1.plot(gate[0]+gate_line[:, 0], gate[1]+gate_line[:, 1], 'kx')
    ax1.plot(gate[0]+gate_line[:, 0], gate[1]+gate_line[:, 1], linestyle='--', color='grey', linewidth=2.5)

# Parameter for the lemniscate
a = np.sqrt(2) * r
b = 2.8

# Generate theta values for parametric equations
theta = np.linspace(0, 2 * np.pi, 1000)

# Parametric equations for the lemniscate
x = (a * np.sqrt(2) * np.cos(theta)) / (1 + np.sin(theta)**2)
y = (b * np.sqrt(2) * np.cos(theta) * np.sin(theta)) / (1 + np.sin(theta)**2)

ax1.plot(y, x)

ax1.annotate('', 
            xy=(1.5, 0), xytext=(0, 0), 
            arrowprops=dict(
                facecolor='red', 
                edgecolor='red', 
                arrowstyle='<|-|>',  # Double-headed arrow
                lw=1,  # Line width to make it thin
                shrinkA=0,  # No shrink on the start
                shrinkB=0   # No shrink on the end
            ))

# Add labels to explain the arrows
ax1.text(0.75
         , 0.15, 'R', ha='center', color='red', fontsize=12)

ax1.annotate('', 
            xy=(-0.3, 1.5), xytext=(-0.3, 0), 
            arrowprops=dict(
                facecolor='red', 
                edgecolor='red', 
                arrowstyle='<|-|>',  # Double-headed arrow
                lw=1,  # Line width to make it thin
                shrinkA=0,  # No shrink on the start
                shrinkB=0   # No shrink on the end
            ))

# Add labels to explain the arrows
ax1.text(-0.50, 0.75, 'R', ha='center', color='red', fontsize=12)

ax1.annotate('', 
            xy=(0.75, 1.8), xytext=(2.25, 1.8), 
            arrowprops=dict(
                facecolor='blue', 
                edgecolor='blue', 
                arrowstyle='<|-|>',  # Double-headed arrow
                lw=1,  # Line width to make it thin
                shrinkA=0,  # No shrink on the start
                shrinkB=0   # No shrink on the end
            ))

# Add labels to explain the arrows
ax1.text(1.5, 2, '1.5 m', ha='center', color='blue', fontsize=12)

# Add circular arrows
arrow1 = FancyArrowPatch(posA=(0.5, 2.5), posB=(-0.5, 2.5),
                        connectionstyle="arc3,rad=0.5",
                        arrowstyle='->',
                        mutation_scale=20,
                        color='black')

ax1.add_patch(arrow1)

# Add circular arrows
arrow2 = FancyArrowPatch(posA=(0.5, -2.5), posB=(-0.5, -2.5),
                        connectionstyle="arc3,rad=-0.5",
                        arrowstyle='->',
                        mutation_scale=20,
                        color='black')

ax1.add_patch(arrow2)
    
ax1.set_xlabel('x (m)')
ax1.set_ylabel('y (m)')
ax1.xaxis.set_ticklabels(['','','',0])
ax1.yaxis.set_ticklabels(['','','','','',0])
ax1.set_aspect('equal', adjustable='box')

plt.savefig('lemniscate_trajectory.eps')
```

```python
test_env = Quadcopter3DGates(
    num_envs=1,
    gates_pos=gate_pos,
    gate_yaw=gate_yaw,
    start_pos=start_pos,
    initialize_at_random_gates=False,
    randomization=randomization_fixed_params_lambda,
    gates_ahead=1,
    num_state_history=0,
    num_action_history=0,
    history_step_size=1,
    param_input=False,
    param_input_noise=0.0
)
```

```python
# MODEL DEFINITION
policy_kwargs = dict(activation_fn=torch.nn.ReLU, net_arch=[dict(pi=[64, 64, 64], vf=[64, 64, 64])], log_std_init = 0)
std_model = PPO(
    "MlpPolicy",
    test_env,
    policy_kwargs=policy_kwargs,
    verbose=0,
    tensorboard_log=None,
    n_steps=1000,
    batch_size=5000,
    n_epochs=10,
    gamma=0.999
)

# std_model = PPO.load('./models/real30/r150/lambda_drone_4/100000000')
std_model = PPO.load('./models/real30/r100/lambda_drone_1/100000000')
```

```python
# TESTING
test_env.reset()

std_position_x = []
std_position_y = []
std_position_z = []
std_speed = []

std_p = []
std_q = []
std_r = []

std_phi = []
std_theta = []
std_psi = []

# do 1000 steps and print state and action
for i in range(1000):
    # print('step', i)
    num = test_env.num_state_history+1
    state_len = int(len(test_env.states[0])/num)

    actions, _ = std_model.predict(test_env.states, deterministic=True)
    states, rewards, dones, infos = test_env.step(actions)

    if (i+1) % 500 == 0: 
        print('step', i+1)
        print('rewards=', rewards)
        print('Num gates passed=', infos)
        print('')
    std_position_x.append(test_env.world_states[0,0])
    std_position_y.append(test_env.world_states[0,1])
    std_position_z.append(test_env.world_states[0,2])
    std_p.append(test_env.world_states[0,9]/ np.pi * 180)
    std_q.append(test_env.world_states[0,10]/ np.pi * 180)
    std_r.append(test_env.world_states[0,11]/ np.pi * 180)
    std_phi.append(test_env.world_states[0,6] / np.pi * 180)
    std_theta.append(test_env.world_states[0,7]/ np.pi * 180)
    std_psi.append(test_env.world_states[0,8]/ np.pi * 180)
    std_speed.append(np.linalg.norm(test_env.world_states[0,3:6]))

# print(f"Max speed: {max(speed)}")
# print(f"Avg speed: {np.mean(speed)}")
# print(f"Min stable speed: {min(speed[500:])}")

np.save("sim_lambda_r100_x.npy", std_position_x)
np.save("sim_lambda_r100_y.npy", std_position_y)
np.save("sim_lambda_r100_z.npy", std_position_z)
np.save("sim_lambda_r100_speed.npy", std_speed)
```
